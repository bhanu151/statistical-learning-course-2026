## Why are the coefficients corresponding to the same features, not same between the Multiple Linear Regression and the Simple Linear Regression?

## Let's try adding some correlation between Hours Studied and Previous Scores

```python
X_with_corr = X.copy()
X_with_corr["Hours Studied Correlated"] = round(X_with_corr["Hours Studied"] - 0.05*X_with_corr["Previous Scores"] + 0.1 * np.random.randn(len(X)))
X_with_corr["Hours Studied Correlated"] = X_with_corr["Hours Studied Correlated"].clip(lower=0)
X_with_corr = X_with_corr.drop(columns=["Hours Studied"])
X_with_corr
```

```python
corr = X_with_corr.corr()
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0)
plt.title("Feature Correlation Matrix")
plt.show()
```

## Simple linear regression after increasing the correlation between hours studied and previous scores

```python
X_single_feature_with_corr = X_with_corr.drop(columns=["Sleep Hours", "Previous Scores", "Extracurricular Activities", "Sample Question Papers Practiced"])
y = data["Performance Index"]
print(X_single_feature.head())

X_train, X_test, y_train, y_test = train_test_split(X_single_feature_with_corr, y, test_size = 0.2, random_state = 42)
print(X_train.shape, y_train.shape, X_test.shape, y_test.shape)
model = LinearRegression()
model.fit(X_train,y_train)
print(f"train score of simple linear regression model with correlation = {model.score(X_train, y_train)}")
print(f"train score of simple linear regression model with correlation = {model.score(X_train, y_train)}")
predict = np.round(model.predict(X_test), decimals = 1)
print(f"test score of simple linear regression model with correlation = {r2_score(y_test,predict)}")
print(f"Model coefficient and intercept = {model.coef_, model.intercept_}")
```

## Multiple linear regression after increasing the correlation between hours studied and previous scores

```python
y = data["Performance Index"]
print(X_with_corr.head())
X_train, X_test, y_train, y_test = train_test_split(X_with_corr, y, test_size = 0.2, random_state = 42)
print(X_train.shape, y_train.shape, X_test.shape, y_test.shape)
model = LinearRegression()
model.fit(X_train,y_train)
print(f"train score of mutiple linear regression model with correlation = {model.score(X_train, y_train)}")
print(f"train score of multiple linear regression model with correlation = {model.score(X_train, y_train)}")
predict = np.round(model.predict(X_test), decimals = 1)
print(f"test score of multiple linear regression model with correlation = {r2_score(y_test,predict)}")
print(f"Model coefficient and intercept = {model.coef_, model.intercept_}")
```

Note how the coefficients corresponding to Hours Studied Correlated are quite different in the two cases.
