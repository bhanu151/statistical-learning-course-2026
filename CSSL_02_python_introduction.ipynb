### Exercise - Calculate the average score of this class

```python
scores = [65, 78, 30, 85]
```

```python
average_score = sum(scores) / len(scores) # Calculate the average
max_score = max(scores) # Gives the maximum
min_score = min(scores) # Gives the minimum
print(f"The class average is {average_score}, and the the class highest is {max_score}")
```

### Functions and if statements

```python
def has_student_passed(score):
  """
  This is a multiline comment.
  This function takes the score and returns pass if
  the score is above 40.
  """
  if score >= 40:
    print("The student has passed the course.") # Gets executed when if condition evaluates to true
  else:
    print("The student has not passed the course.") # Gets executed when if condition evaluates to false
  return
```

### For loop

```python
for s in scores:
  print(f"s={s},")
  has_student_passed(s)
```

## Homework: Write a function to calculate grades and return the grade

```python
# The function will look something like this

def get_grade(score):
  """
  This is a multiline comment.
  This function takes the score and returns the grade.
  """
  if score >= 40 and score <=50:
    return "E"
  elif score>50 and score<=60:
    return "D"
  elif score>60 and score<=100:
    return "A"
  return "U"
```
